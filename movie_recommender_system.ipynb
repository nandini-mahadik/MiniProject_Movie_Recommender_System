```python
#creating helper function
def convert(obj):
    L = []
    for i in ast.literal_eval(obj): #converts the object into a list
        L.append(i['name'])
    return L
```

```python
#here is one problem that the above list is a string 
#and when we put that list into a funtion it will create an error like this
convert('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
#here the names are in string format 
#to solve this we will convert the string of list into a list
```

```python
#converting string of list into a list
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]') 
#ast is a inbuilt module in python and the function literal_eval() does the same work
```

```python
#here we are acessing only top 3 casts of the movie
#from each dictionary we want the actual name of the cast 
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
#so here overview is in string format we will convert it into a list
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
#Sam Worthington => SamWorthington
#Science Fiction => ScienceFiction
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywoeds'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
import nltk 
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
#now convert the entire string into a lower case
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
```
